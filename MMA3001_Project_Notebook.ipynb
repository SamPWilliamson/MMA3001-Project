# 1. 📘 Project Introduction
This section introduces the real-world engineering context, the energy crisis in educational facilities, the failure modes of privacy-invasive optical and motion sensors, the governing physics of room air dynamics, and the formal statement of AI assistance.


## 1.1 🏛️ The Real-World Engineering Problem
Heating, Ventilation, and Air Conditioning (HVAC) accounts for **approximately 40% of all energy consumed by educational and commercial infrastructure**. 

In conventional Building Management Systems (BMS), ventilation fans and chillers operate on static, "dumb" time schedules (for example, forcing 100% fan speed between 08:00 and 18:00 Monday to Friday). This creates severe operational inefficiencies:
* **Over-ventilation & Energy Waste:** An empty 100-person lecture theatre is continuously heated, cooled, and ventilated at maximum load, consuming hundreds of unneeded kilowatt-hours.
* **Under-ventilation & Cognitive Fatigue:** When an unscheduled workshop or design sprint packs 60 students into an unventilated classroom, indoor Carbon Dioxide ($CO_2$) rapidly spikes above $1,200\text{--}1,600\text{ ppm}$, causing lethargy, reduced cognitive performance, and elevated airborne pathogen transmission.

> **🎯 The Engineering Goal:**  
> Infer real-time room occupancy count $N(t)$ using non-invasive, low-cost environmental telemetry (indoor $CO_2$, temperature, relative humidity) to enable **Demand-Controlled Ventilation (DCV)**, whilst completely eliminating cameras and motion sensors.


## 1.2 🚫 Why Cameras and PIR Sensors Fail
Modern facilities cannot simply install cameras or motion detectors across every room due to three engineering constraints:
1. **Privacy & Ethical Liabilities:** Optical video surveillance in classrooms creates severe privacy friction, GDPR compliance risks, and cybersecurity vulnerabilities.
2. **PIR Sensor Blindness:** Passive Infrared (PIR) motion detectors only register moving thermal gradients. When 40 engineering students sit still during a 2-hour examination or theoretical lecture, PIR sensors assume the room is completely unoccupied and throttle fresh air!
3. **High Deployment Cost & Obstructions:** Stereoscopic vision cameras require expensive infrastructure, regular optical calibration, and suffer from visual blind spots behind acoustic partitions and architectural pillars.

Environmental IoT sensors ($CO_2$, Temperature, Humidity) are non-invasive, preserve 100% human privacy, are cost-effective, and capture whole-room metabolic exhalation.


## 1.3 🪣 The Physics Challenge: "The Leaky Bucket"
Why does a standard, naive machine learning model fail when given raw $CO_2$ values?

Indoor air behaves dynamically like a **leaky bucket**:
* **Occupants breathing** acts like water pouring in from a tap ($G \cdot N(t)$).
* **HVAC mechanical ventilation** acts like a drain hole at the bottom flushing air out ($Q(t) \cdot (C(t) - C_{\text{amb}})$).

From first-principles conservation of chemical mass, the transient rate of change of $CO_2$ inside room volume $V$ is governed by:
$$V \frac{dC}{dt} = G \cdot N(t) - Q(t) \cdot \left(C(t) - C_{\text{amb}}\right)$$

Where:
* $V \approx 450\text{ m}^3$: Effective room volume of Room G.57 in cubic metres.
* $C(t)$: Instantaneous indoor $CO_2$ concentration ($\text{ppm}$).
* $C_{\text{amb}} \approx 415\text{ ppm}$: Outdoor background $CO_2$ baseline.
* $G \approx 0.005\text{ L/s}$: Average per-person metabolic $CO_2$ exhalation rate.
* $Q(t)$: Mechanical and infiltration ventilation exchange rate ($\text{m}^3/\text{s}$).

Rearranging for instantaneous occupant headcount $N(t)$:
$$\mathbf{N(t) = \frac{1}{G} \left[ V \frac{dC}{dt} + Q(t) \cdot (C(t) - C_{\text{amb}}) \right]}$$

> [!IMPORTANT]
> **The Critical Physics Insight:**  
> Occupancy count $N(t)$ is directly proportional to the **rate of change (the first derivative $\frac{dC}{dt}$)**, not merely the absolute concentration $C(t)$!  
> Because of room volume, $CO_2$ exhibits a **20 to 40 minute pneumatic lag**. When 30 students enter, $CO_2$ remains near $450\text{ ppm}$ initially, but $\frac{dC}{dt}$ surges upward immediately. When students leave, $CO_2$ remains high ($800\text{ ppm}$) whilst slowly diluting, but $\frac{dC}{dt}$ flips negative.  
> A naive model without numerical derivatives achieves an unacceptable $R^2 = 0.13$. Injecting first-principles numerical derivatives boosts predictive performance to $R^2 = 0.48$ (+271% gain).

## 4.2 🧮 First-Principles Numerical Methods Implementation
Below, we implement our core numerical algorithms directly from first principles using pure NumPy to verify mathematical rigor and compliance with unit syllabus topics.

### 4.2.1 2nd-Order Central Difference Derivative (Week 7)
Using Taylor series expansions:
$$f(x+h) = f(x) + hf'(x) + \frac{h^2}{2}f''(x) + \frac{h^3}{6}f'''(x) + \mathcal{O}(h^4)$$
$$f(x-h) = f(x) - hf'(x) + \frac{h^2}{2}f''(x) - \frac{h^3}{6}f'''(x) + \mathcal{O}(h^4)$$

Subtracting the two equations cancels all even-order terms, yielding the central difference formula:
$$f'(x) = \frac{f(x+h) - f(x-h)}{2h} - \frac{h^2}{6}f'''(\xi) = \frac{f(x+h) - f(x-h)}{2h} + \mathcal{O}(h^2)$$

### 4.2.2 Composite Simpson's 1/3 Rule (Week 6)
Approximates the integrand by piecewise quadratic polynomials over adjacent pairs of intervals:
$$\int_{a}^{b} f(x)\, dx = \frac{h}{3}\left[f(x_0) + 4f(x_1) + 2f(x_2) + 4f(x_3) + \dots + 4f(x_{n-1}) + f(x_n)\right] - \frac{(b-a)h^4}{180}f^{(4)}(\xi)$$
The global truncation error scales as $\mathcal{O}(h^4)$.

### 4.2.3 Natural Cubic Spline Interpolation with Thomas Algorithm (Week 3)
Constructs cubic polynomials $S_i(x) = a_i + b_i(x-x_i) + c_i(x-x_i)^2 + d_i(x-x_i)^3$ enforcing $C^0, C^1,$ and $C^2$ continuity across all interior nodes, solved via the $O(n)$ Thomas tridiagonal matrix algorithm.


````python
# Cell 4.2: First-Principles Numerical Methods Library & Analytical Verification
from typing import Optional, Tuple

def central_difference_derivative(x: np.ndarray, y: np.ndarray, smooth_window: Optional[int] = None) -> np.ndarray:
    """Compute first derivative using 2nd-order central differences with endpoint handling.
    
    Formula:
        Interior: f'(x_i) = (f(x_{i+1}) - f(x_{i-1})) / (2h) + O(h^2)
        Boundaries: 2nd-order 3-point asymmetric formulas: f'(x_0) = (-3f_0 + 4f_1 - f_2)/(2h) + O(h^2)
    """
    x = np.asarray(x, dtype=np.float64)
    y = np.asarray(y, dtype=np.float64)
    n = len(x)
    if n < 2:
        raise ValueError("At least 2 points required.")
    
    # Optional pre-smoothing to prevent O(epsilon/h) noise explosion
    if smooth_window is not None and smooth_window >= 3:
        w = int(smooth_window) | 1
        pad = w // 2
        y_padded = np.pad(y, pad, mode="edge")
        y_calc = np.convolve(y_padded, np.ones(w)/w, mode="valid")
    else:
        y_calc = y.copy()
        
    dydx = np.zeros(n, dtype=np.float64)
    dx = np.diff(x)
    h = dx[0]
    
    # Vectorised interior central differences
    dydx[1:-1] = (y_calc[2:] - y_calc[:-2]) / (2.0 * h)
    # Second-order 3-point single-sided differences at boundaries (matching Report Section 3.2)
    if n >= 3:
        dydx[0] = (-3.0 * y_calc[0] + 4.0 * y_calc[1] - y_calc[2]) / (2.0 * h)
        dydx[-1] = (3.0 * y_calc[-1] - 4.0 * y_calc[-2] + y_calc[-3]) / (2.0 * h)
    else:
        dydx[0] = (y_calc[1] - y_calc[0]) / h
        dydx[-1] = (y_calc[-1] - y_calc[-2]) / h
    return dydx


def composite_simpsons_rule(x: np.ndarray, y: np.ndarray) -> float:
    """Compute definite integral using Composite Simpson's 1/3 Rule with O(h^4) error."""
    x = np.asarray(x, dtype=np.float64)
    y = np.asarray(y, dtype=np.float64)
    n = len(x) - 1
    if n < 2:
        raise ValueError("At least 3 points (2 intervals) required for Simpson's rule.")
    
    h = (x[-1] - x[0]) / n
    if n % 2 == 0:
        # Standard Composite Simpson's 1/3 Rule
        integral = y[0] + y[-1] + 4.0 * np.sum(y[1:-1:2]) + 2.0 * np.sum(y[2:-1:2])
        return float(integral * (h / 3.0))
    else:
        # Simpson's 1/3 for first n-1 intervals + Trapezoid rule for final interval
        integral = y[0] + y[-2] + 4.0 * np.sum(y[1:-2:2]) + 2.0 * np.sum(y[2:-2:2])
        simp_part = integral * (h / 3.0)
        trap_part = 0.5 * h * (y[-2] + y[-1])
        return float(simp_part + trap_part)


# --- Analytical Verification & Convergence Check ---
print("Running Analytical Convergence Verification...")

# Test 1: Verify O(h^2) derivative error scaling on f(t) = sin(2*pi*t)
t_fine = np.linspace(0, 1, 1001)
errors = []
step_sizes = [0.1, 0.05, 0.025, 0.0125]

for h_step in step_sizes:
    t_test = np.arange(0, 1.0 + h_step, h_step)
    y_test = np.sin(2.0 * np.pi * t_test)
    exact_deriv = 2.0 * np.pi * np.cos(2.0 * np.pi * t_test)
    num_deriv = central_difference_derivative(t_test, y_test)
    # Check max interior error
    err = np.max(np.abs(num_deriv[1:-1] - exact_deriv[1:-1]))
    errors.append(err)

print("Derivative Error vs Step Size (Testing O(h^2) convergence):")
for i in range(len(step_sizes)):
    ratio = errors[i-1]/errors[i] if i > 0 else 1.0
    print(f" - h = {step_sizes[i]:.4f}: Max Error = {errors[i]:.6e} (Error reduction ratio: {ratio:.2f}x ~ expected 4.0x)")

# Test 2: Verify O(h^4) Simpson's integration on f(t) = t^3 from 0 to 2 (Exact = 2^4/4 = 4.0)
t_quad = np.linspace(0, 2, 21) # 20 intervals
y_quad = t_quad**3
num_int = composite_simpsons_rule(t_quad, y_quad)
print(f"\nSimpson's 1/3 Rule on Integral[t^3 dt, 0..2]: Calculated = {num_int:.8f}, Exact = 4.00000000 (Error = {abs(num_int - 4.0):.2e})")
print("✅ First-principles numerical routines successfully verified.")
````

## 4.4 🏆 Model Training & Benchmark Comparison Suite
We fit and evaluate four distinct machine learning architectures:
1. **Baseline KNN ($K=5$):** Naive non-parametric benchmark trained solely on raw $CO_2$ and time.
2. **Pruned Decision Tree ($\text{max\_depth}=5$):** Ultra-low-latency rule-based model ideal for microcontrollers.
3. **Gaussian Process Regressor (RBF Kernel):** Probabilistic Bayesian regression producing calibrated uncertainty bounds ($\mu \pm 1.96\sigma$).
4. **Random Forest Ensemble ($B=50$ Trees):** Variance-reduced bagging ensemble.


````python
# Cell 4.4: Model Training, Inference & Scorecard Generation
import time
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, WhiteKernel
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

models = {}
predictions = {}
timing = {}

# 1. Baseline KNN (K=5)
print("Training Baseline KNN (Raw CO2 + Time)...")
t0 = time.perf_counter()
knn = KNeighborsRegressor(n_neighbors=5, weights="distance")
knn.fit(X_train_base, y_train)
t_fit = time.perf_counter() - t0
t1 = time.perf_counter()
pred_knn = np.clip(knn.predict(X_test_base), 0, 100)
t_pred = (time.perf_counter() - t1) / len(X_test_base) * 1e6
models["Baseline KNN"] = knn
predictions["Baseline KNN"] = pred_knn
timing["Baseline KNN"] = t_pred

# 2. Pruned Decision Tree (depth=5)
print("Training Pruned Decision Tree (Full Physics Features)...")
t0 = time.perf_counter()
dt = DecisionTreeRegressor(max_depth=5, min_samples_leaf=10, random_state=42)
dt.fit(X_train_full, y_train)
t1 = time.perf_counter()
pred_dt = np.clip(dt.predict(X_test_full), 0, 100)
t_pred = (time.perf_counter() - t1) / len(X_test_full) * 1e6
models["Decision Tree"] = dt
predictions["Decision Tree"] = pred_dt
timing["Decision Tree"] = t_pred

# 3. Gaussian Process Regressor (Trained on representative subset for matrix inversion feasibility)
print("Training Gaussian Process Regressor (RBF Kernel)...")
gpr_subsample_size = min(1500, len(X_train_full))
np.random.seed(42)
gpr_indices = np.random.choice(len(X_train_full), gpr_subsample_size, replace=False)
kernel = RBF(length_scale=1.0) + WhiteKernel(noise_level=0.1)
gpr = GaussianProcessRegressor(kernel=kernel, alpha=1e-2, normalize_y=True, random_state=42)
gpr.fit(X_train_full[gpr_indices], y_train[gpr_indices])
t1 = time.perf_counter()
pred_gpr, pred_gpr_std = gpr.predict(X_test_full, return_std=True)
pred_gpr = np.clip(pred_gpr, 0, 100)
t_pred = (time.perf_counter() - t1) / len(X_test_full) * 1e6
models["Gaussian Process"] = gpr
predictions["Gaussian Process"] = pred_gpr
timing["Gaussian Process"] = t_pred

# 4. Random Forest Ensemble (B=50)
print("Training Random Forest Ensemble (B=50)...")
rf = RandomForestRegressor(n_estimators=50, max_depth=10, min_samples_leaf=5, n_jobs=-1, random_state=42)
rf.fit(X_train_full, y_train)
t1 = time.perf_counter()
pred_rf = np.clip(rf.predict(X_test_full), 0, 100)
t_pred = (time.perf_counter() - t1) / len(X_test_full) * 1e6
models["Random Forest"] = rf
predictions["Random Forest"] = pred_rf
timing["Random Forest"] = t_pred

# Compute Scorecard Metrics
operating_mask = test_df["is_operating_hours"].values == 1
actual_occupied = y_test > 0
critical_mask = operating_mask & actual_occupied

scorecard_rows = []
for name, preds in predictions.items():
    r2 = r2_score(y_test, preds)
    mae = mean_absolute_error(y_test, preds)
    rmse = float(np.sqrt(mean_squared_error(y_test, preds)))
    
    # Operational False Negative Rate: Predict 0 occupants when room is occupied during working hours
    pred_zero = preds < 0.5
    fn_count = np.sum(critical_mask & pred_zero)
    fnr = (fn_count / np.sum(critical_mask)) * 100.0 if np.sum(critical_mask) > 0 else 0.0
    
    scorecard_rows.append({
        "Model Architecture": name,
        "Input Features": "Raw CO2 + Time" if "Baseline" in name else "Full Physics (16)",
        "Test R^2": round(r2, 4),
        "MAE (people)": round(mae, 2),
        "RMSE (people)": round(rmse, 2),
        "Working Hours FNR (%)": f"{fnr:.2f}%",
        "Inference Latency (us)": f"{timing[name]:.1f} us"
    })

scorecard_df = pd.DataFrame(scorecard_rows)
print("\n" + "="*80)
print("🏆 FINAL BENCHMARK SCORECARD (HELD-OUT TEST SET)")
print("="*80)
display(scorecard_df)
````

* **Question:** *"Why select a simple Decision Tree over the more sophisticated Gaussian Process or Random Forest?"*
* **Defensible Response:** In HVAC control, failing to ventilate an occupied classroom (False Negative) impairs student health. Both the Decision Tree and Random Forest achieved a **$0.00\%$ False Negative Rate** during working hours. However, the Decision Tree executes in only **$150\,\mu\text{s}$** with **$5.8\text{ KB}$ RAM** (vs $23,300\,\mu\text{s}$ for Random Forest), making it the only model deployable on low-cost ARM Cortex-M microcontrollers at the edge.
