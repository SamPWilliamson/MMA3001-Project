## 📂 1.3.1 Why You **Need** to Document Your Code

---

Take a look at the examples of code you have seen so far in this unit. Now imagine all you had was the code - no text explanation about what it was trying to do and no comments on any line.

For example, consider the following block of code.

```python
import numpy as np
from scipy.optimize import minimize
def f(x):
    return x**4 - 2*x**3 + 4*x**2 + 5
a = 0.5
b = minimize(f, a)
print(b.x[0])
print(b.fun)
```

Even with a *relatively* readable language, like Python, it will probably take you a while to work out what is going on here. Consider the same code, present below.

```python
# import numpy library and scipy minimize function
import numpy as np
from scipy.optimize import minimize

# Define a polynomial function
def polynomial_function(x):
    return x**4 - 2*x**3 + 4*x**2 + 5

# Initial guess for the minimum
x0 = 0.5

# Find the local minimum
result = minimize(polynomial_function, x0)

# Print the results
print(f"Local minimum found at x = {result.x[0]:.4f}")
print(f"Function value at local minimum = {result.fun:.4f}")
```
The addition of a couple of short comments and more meaningful variable names (e.g. x0 vs a), make the code a lot easier to follow.

Now you can hopefully understand what the value of 'a' is used for and the purpose of the code, i.e. to find the minimum of a polynomial function. If you wanted to modify the code to change the function, this would now be a much easier job.

Providing good documentation for your code:
* Provides **Clarity**, helps provide context and purpose to a codes and algoritms on which it is  based.
* Simplifies **Maintenance**, makes it easier to implement changes, fix bugs and add new features.
* Aids **Collaboration**, helps others understand changes you have made and how a function should be called and the output to expect. Highlights the limits of present routines or flags part of the code that still requires testing or may have room for improvement.
* Speeds up **Onboarding**, makes it easier for new users or developers to understand how a code is structured and how it works.
* Helps **Compliance** and **Review** of code by others and helps them to determine if the code follows appropriate engineering methods and assumptions.

It also really helpful when you move onto a new project and realise you wrote some code to process some data a couple of years ago that could be used again on a similar dataset. If only you could remember where you put it (hopefully in your GitHub) and could recall how to run it (comments and README file).

## 📂 1.3.2 Different Types of Documentation

---

There are multiple types of code documentation, which serve different purposes and in some cases different audiences. All help to improve the usability and value of any code you generate. The following is not an extensive list, but it will highlight the bare minimum that you should include in any code you are required to develop.

> ### ⚠️ Point of note
> We will expect to see evidence of professional documentation in your Project for this unit.

### 🔹 In-line Code Documentation

These, also referred to as *code comments*, are plain text comments that sit within the source code itself. In Python these are prefaced with the '#' symbol and commonly rendered as <font color="green">green</font> text in most code editors (these can be normally be changed in setting if you like).

> ### ⚠️ Point of note
> These act as reminders to developers and those trying to understand a code.

### 🔹 Documentation Strings (Docstrings)

Docstrings are special comments that provide information about functions and classes that should follow a standard form. In Python, these comments are stored as part of a function as a `.__doc__` that attaches to the function and can be called at runtime.

> ### ⚠️ Point of note
> These are comments intended to help users of a code or function without them needing to see the source code.

Docstrings begin and end with triple quotes """, and can span multiple lines.
A docustring for the polynomial function in the last example should look something like this.



````python
def polynomial_function(x):
    """
    Evaluate the  polynomial function $y = x^4 - 2x^3 + 4x^2 + 5$

    Parameters
    ----------
    x : int or float
     the point at which to evaluate the polynomial.

    Returns
    -------
    return : int or float
     the value of the polynomial evaluated at x.
    """
    return x**4 - 2*x**3 + 4*x**2 + 5
````

> ### 🧩 Code Explanation
> This is a NumPy Style Docstring, which sets out the purpose of the function:
>
>`Evaluate the  polynomial function $y = x^4 - 2x^3 + 4x^2 + 5$`
>
> Explains the input parameters, there types and what they represent
>```
>Parameters
>    ----------
>    x : int or float
>     the point at which ...
>```
>
> and the data that will be returned by the function
> ```
>Returns
>    -------
>    return : int or float
>      the value of ...

What sets this aside from normal *inline comments*, is that a user can query a function and return this information using the *help* function in Python.

Let's see what happends when we run `help` on the function we just defined.

````python
help(polynomial_function)
````

#### 🧱 Sphinx / reStructuredText (reST)
```
def add(a, b):
    """
    Add two numbers.

    :param a: First number to add.
    :type a: float
    :param b: Second number to add.
    :type b: float
    :returns: The sum of ``a`` and ``b``.
    :rtype: float
    :raises TypeError: If either argument is not numeric.
    """
    return a + b
```

#### 🧱 Google Style
```
def add(a, b):
    """
    Add two numbers.

    Args:
        a (float): First number to add.
        b (float): Second number to add.

    Returns:
        float: The sum of a and b.

    Raises:
        TypeError: If either argument is not numeric.
    """
    return a + b
```

#### 🧱 NumPy Style
```
def add(a, b):
    """
    Add two numbers.

    Parameters
    ----------
    a : float
        First number to add.
    b : float
        Second number to add.

    Returns
    -------
    float
        The sum of a and b.

    Raises
    ------
    TypeError
        If either argument is not numeric.
    """
    return a + b
```

> ### 💬 While this is subjective, I personally like the **Google Style** as it's easier to read than Sphinx and is a bit more concise than the NumPy Style.

## 📂 1.3.3 Flow Charts and Algorithms
---

Flow charts provide a graphical representation of a specific sequence of steps that a code follows to complete a task. They provide a useful tool in planning a code, and in outlining the steps that an algorithm must follow. They are also independent of programming language and can help explain logic without getting confused by Python- / C++ or Fortran-specific syntax.

When working collaboratively, a flow chart can also help define the inputs and outputs that might be expected from each part of the code. This makes it is easier to segregate and divide work between users, ensuring they understand how their part must fit within the greater whole.

To help users interpret these diagrams, it pays to keep to a standard form. There are different standards for documents, data flows and programs, but also a fair bit of commonality. The most common symbols are as follows:

In some cases, you may see an ellipse used in place of an Oval (rounded rectangle), but it will similarly be used to mark the start and end of a process.

As processes get larger and more complex, it can become necessary to spread them across multiple pages and charts, just as it becomes good practice to split code across multiple functions and .py files. In this case additional symbols are introduced.

* A **small circle** with any number or letter in it, is used as a connector symbol. This allows you to connect one part of a flow-chart with sections that are far away or on a separate page.
* A **rectangle with double vertical sides** is used to denote a subprocess. When this box is encountered, the flow goes to the subroutine, and it continues until a *Return* statement is encountered. Then it goes back to the main flow chart and the flow resumes.

### 🔹 Digitally Generating Flow Charts

There is no limit to how you can generate flow charts. You might sketch these on a whiteboard in a meeting, drag and drop symbols in PowerPoint or Illustrator etc.

Given you are going to be the most technologically literate Engineering Grads we have graduated in the last decade, and with the benefit of AI will be able to produce high-quality work faster than previous generations - let's look at a neat way to do this by introducing the Mermaid diagrams.

> ### 📖 **Mermaid**
is a text-based, Markdown-inspired scripting language that automatically generates and renders diagrams and charts using JavaScript. A few key benefits of this are:
> * A relatively *Simple Syntax*: Uses readable text to define nodes and connections (e.g., A --> B).
> * *Multiple Types*: Supports flowcharts, Gantt charts, pie charts, and mind maps, etc.
> * *Integration:* Works natively in GitHub and Jupyter notebooks.
> * *Benefits:* Solve "doc-rot" (outdated documentation) as diagrams can easily updated alongside code.
> * *AI Compatible:* AI tools can easily generate code and even translate Python code into a Mermaid script.

> ### 📖 **Doc-Rot**
 refers to the gradual and too common process where documentation becomes *outdated*, and subsequently inaccurate, as the code or processess evolve. Everyone wants to create, but few want to document the process.
>
> It *can be worse* than having no documentation as it leads to a loss of trust in the documentation and can be misleading.

### 🧪 Mermaid Flow Chart Example

As an example, let's reproduce the basic flow chart from Figure 1.3.3.

To do this using the *colab_print* library, we first need to install this in our current Colab *runtime environment*.

> ### ⚠️ Point of note
> Anything changes we make to our runtime will reset whenever the runtime disconnects. If you get an error, you may just need to repeat commands from earlier cells. You can use Colab's `Run all` button if you want to automate this.

Given this is the first time we are doing this, I will explain the installation part of the code before we get to the Mermaid part.

> ### 🧩 Code Explanation
> You can install a large number of Python libraries using the `pip install` command.
> to avoid running this installation each time we run the cell, I have embedded the `pip` commands inside a `try` and `except`.
>
> Each time the cell runs, it will try to execute `from colab_print import mermaid`, if this fails, rather then display and error and stopping, the cell will execute the `except` code, causing it to install the missing library.

Let's run it and see what happens.

````python
# Try to access the colab_print tool
try: from colab_print import mermaid
except:
  # access failed to install it and try again
  !pip install colab-print
  from colab_print import mermaid

# run the mermaid tool -
# NOTE: you can't include # comments inside the mermaid script
mermaid('''
flowchart TD;
    A([Start]) --> B[Step 1]
    B --> C{Decision}
    C -- No --> D[Step 2]
    C --- YES --> E([End])
''',theme='neutral')
````

The first time you run this you will see a list of lines that correspond to Colab finding, downloading and installing the colab-print library. Next time you run it, this will not be necessary (at least until the *runtime* resets).

Now let's look at the rest of the code.

>### 🧩 Code Explanation
> `mermaid(''' ''')` is a call to the mermaid function in the *colab-print* library. This will pass anything between the three apostrophes as ''' mermaid script '''.

The Mermaid script starts by specifying the chart type, i.e. `flowchart TD`, with *TD* used to denote a Top-Down chart flow.

The letters, A,B, etc. are used to represent objects or nodes in the diagram, with --> used to create pathways between them.

`A --> B` means create a path from node A to node B.

The (), [], or {} brackets after each letter are used to define the node type and the -- No and the --- YES, before the --> adds the condition to the flow branch. The extra - before YES is optional, but added here to make the line with YES longer than the line for No. The *theme* controls the colour scheme and appearance.

You can probably get a better sense of this from the Video where we look at the *Mermaid.Live* web-based example. The advange of doing it via the Colab cell is that we get an editable version of our chart right here in our notebook.

## 📂 1.3.4 Tools for Generating Documentation

---

The term ["doc-rot"](#scrollTo=tpg8r7VseZzv&line=13&uniqifier=1) was mentioned above in regards to flow charts, but it is a common problem, as development tends outpace documentation. Fortunately, when it comes to software, there is a range of tools that can be used to aid this process and remove many of the hurdles to effective documentation.

The simplest of these will go through your code and extract a list of all the functions and how they're interconnected. If you have diligently included appropriate **docstrings**, these tools will use this information to provide context to these different parts of the code.

The easiest tool for this is probably **pdoc**, but there are other packages that work across multiple languages and give greater control of the output documentation - which may be necessary for larger and more complex projects.

> ### ⚠️ Point of note
> **pdoc** is NOT **pydoc**
> * pydoc - a built-in Python module that serves as an offline reference manual and basic documentation generator,
> * pdoc - a third-party library designed specifically to auto-generate modern, visually appealing API documentation web pages.
