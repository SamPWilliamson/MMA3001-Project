## 📂 10.4.1 How Ordinary Differential Equations Relate to Partial Differential Equations

---

In week 9, we focused on the numerical solution of **Ordinary Differential Equations (ODEs)**. Most of what we discussed for ODEs can be adapted to the solution of **Partial Differential Equations (PDEs)**.

**PDEs** arise when a system's state depends on multiple independent variables (e.g., both time and spatial dimensions), extending the concept of ODEs to higher-dimensional scenarios.

>The moment your state depends on where as well as when, ordinary derivatives split into partial derivatives, and curves become surfaces.

In essence:
* **ODE**: one independent variable → solution is a curve (could be space or time)
* **PDE**: multiple independent variables → solution is a surface/field.
This means:
    - More derivatives (∂/∂t, ∂/∂x, ∂²/∂x², …)
    - More conditions (initial + boundary)
    - More equation types (elliptic, parabolic, hyperbolic)
    - More numerical complexity (grids, stencils, stability constraints)

The study of ODEs provides a good grounding before moving to the more complex PDEs.

PDEs determine how almost every system operates in *"space"* and *"time"*. As such, they sit at the heart of mechanical and aerospace engineering because they are the mathematical language of continuum physics.

Whenever an engineer models how a system moves, deforms, vibrates, heats, or flows, they are ultimately describing how a physical field evolves in space and time — and that evolution is governed by partial differential equations.

In mechanics, PDEs capture the behaviour of stress, strain, and displacement fields in solids, enabling the analysis of structural integrity, fatigue, buckling, and vibration modes in everything from turbine blades to automotive components. In aerospace engineering, PDEs underpin the governing equations of fluid dynamics and thermodynamics — the Navier-Stokes equations for airflow around wings and fuselages, the heat equation for thermal protection systems, and wave equations for aeroelasticity and flutter. These equations allow engineers to predict lift, drag, shock formation, boundary-layer behaviour, thermal loads, and dynamic stability. In both fields, PDEs provide the bridge between fundamental physical laws and the computational tools — finite elements, finite volumes, CFD, and multiphysics simulation — that make modern engineering design possible.

## 📂 10.4.2 Definition of a Partial Differential Equation

---

> ### 📖 Definition
> A partial differential equation (PDE) is an equation that relates an unknown function of multiple independent variables (e.g., $u(x,y,z,t)$ where $x,y,z,t$ are the **independent variables**) to its partial derivatives $\left(\mbox{e.g., } \frac{\partial}{\partial x},\frac{\partial}{\partial y},\frac{\partial}{\partial z},\frac{\partial}{\partial t}\right)$. In other words, it describes how a field—a quantity that varies over space, time, or both—changes in response to physical laws.

Common examples in engineering include the one-(or higher)-dimensional wave equation
$$\frac{\partial^2u}{\partial t^2} = c^2\frac{\partial^2u}{\partial x^2},$$

where $c$ is the wave speed, governs how a disturbance, $u(x,t)$ (displacement or pressure change) propagates through a medium.

The diffusion equation we used for heat conduction through a rod in Notebook 10.3,
$$ \frac{\partial u}{\partial t}=\alpha \frac{\partial^2 u}{\partial x^2},$$
or the *Poisson equation*,
$$ \frac{\partial^2 u}{\partial x^2}+\frac{\partial^2 u}{\partial y^2}=\rho(x,y),$$
which represents a broad range of processes from steady-state heat conduction to how pressure adjusts to ensure incompressibility in a fluid flow.

Let's look at how we can classify a PDE and determine its order.
1. If a function has only one independent variable (e.g., $u(t)$), it is an **ODE**. If it has more than one independent variable (e.g., $u(x,t)$), then it is a **PDE**.
2. Identify the highest derivative, which determines the order of the system.

##  🧠 10.4.3 Classification of Partial Differential Equations
---

There are three types of PDEs: *hyperbolic*, *parabolic*, and *elliptic*. This is **important** as the numerical methods for solving these types are different. The exact mathematical proof of their behaviour is beyond the scope of this unit, but as an *engineer*, you need to be able to identify one type from the other and understand what this means for their solution.

For a second-order PDE, you can do this as follows:
1. Write the PDE out in the form
$$A\frac{\partial^2 \psi}{\partial x^2} + B\frac{\partial^2 \psi}{\partial x \partial y}+C\frac{\partial^2 \psi}{\partial y^2}+D\frac{\partial \psi}{\partial x} + E\frac{\partial \psi}{\partial y}+F\psi +G=0$$
2. Consider the coefficients $A,B,C$ as defining a surface in multidimensional space via the quadratic term
$$A x^2 + B xy + Cy^2 = \mbox{const.}$$
3. Determine the *discriminant*
$$ \Delta = B^{2} - 4AC$$

If:
* $\Delta > 0$ - the equation is **hyperbolic**;
* $\Delta = 0$ - the equation is **parabolic**;
* $\Delta < 0$ - the equation is **elliptic**;

**Note**: $A,B,C$ are the dominant terms that dictate how the equation behaves. The lower-order terms $D, E, F, G$ can be neglected for classification.

### 🔹 PDE Classification via Discriminant $\Delta = B^2 - 4AC $

| Discriminant Condition | PDE Type     | Conic Section | Characteristics                         | Behaviour / Information Transport                                                                 |
|------------------------|--------------|---------------|------------------------------------------|----------------------------------------------------------------------------------------------------|
| $$ \Delta < 0 $$       | **Elliptic** | Ellipse       | No real characteristics                  | Smooth solutions; global influence; no propagation; interior values depend on the entire boundary |
| $$ \Delta = 0 $$       | **Parabolic**| Parabola      | One repeated characteristic direction    | Diffusion; smoothing; infinite spatial propagation speed; preferred time direction                |
| $$ \Delta > 0 $$       | **Hyperbolic**| Hyperbola     | Two real characteristic families         | Finite‑speed propagation; wavefronts; causal cones; signals travel along two characteristic paths     |

---

Characteristic paths are trajectories along which a PDE can be reduced to a system of first-order ODEs.

This has significant implications for the numerics of these PDEs.

### 💬 Aside: Characteristic Lines and Information Propagation in a PDE (Optional)

The section above is sufficient for our purposes in this unit, but if you want to understand why the elliptic equation behaves differently from the parabolic equation, you need to consider the *characteristic lines*. As discussed above, these are paths along which a PDE can be reduced to a system of first-order ODEs.

Take a PDE of the form
$$A\frac{\partial^2 \psi}{\partial x^2} + B\frac{\partial^2 \psi}{\partial x \partial y}+C\frac{\partial^2 \psi}{\partial y^2}
= f\left(x,y,\psi,\frac{\partial \psi}{\partial x}, \frac{\partial \psi}{\partial y}\right),\tag{10.8}$$

where the remaining lower order terms are absorbed into the right-hand side.

To break the second-order derivatives down to first-order, we define the variables
$$p=\frac{\partial \psi}{\partial x},\qquad q=\frac{\partial \psi}{\partial y},$$

Substituting these into the PDE yields,

$$A\frac{\partial p}{\partial x} + B\frac{\partial p}{\partial y}+C\frac{\partial q}{\partial y}
= f\left(x,y,\psi,\frac{\partial \psi}{\partial x}, \frac{\partial \psi}{\partial y}\right)\tag{10.9},$$

We can represent the change in $\psi$ and the introduced variables $p$ and $q$ via the *chain-rule*, i.e.
\begin{align}
 \mbox{d}\psi &= \frac{\partial \psi}{\partial x}\mbox{ d}x+\frac{\partial \psi}{\partial y}\mbox{ d}y = p\mbox{ d}x + q\mbox{ d}y\tag{10.10}\\
 \mbox{d}p &= \frac{\partial p}{\partial x}\mbox{ d}x+\frac{\partial p}{\partial y}\mbox{ d}y = \frac{\partial^2 \psi}{\partial x^2}\mbox{ d}x + \frac{\partial^2 \psi}{\partial x\partial y}\mbox{ d}y \tag{10.11}\\
 \mbox{d}q &= \frac{\partial q}{\partial x}\mbox{ d}x+\frac{\partial q}{\partial y}\mbox{ d}y = \frac{\partial^2 \psi}{\partial x \partial y}\mbox{ d}x + \frac{\partial^2 \psi}{\partial y^2}\mbox{ d}y \tag{10.12}
\end{align}

We can rearrange equations 10.11 and 10.12 to give the following relationships for $\frac{\partial^2 \psi}{\partial x^2}$ and $\frac{\partial^2 \psi}{\partial y^2}$,

$$\frac{\partial^2 \psi}{\partial x^2} = \frac{\mbox{d}p-\frac{\partial^2 \psi}{\partial x\partial y}\mbox{ d}y}{\mbox{d}x}$$

$$\frac{\partial^2 \psi}{\partial y^2} = \frac{\mbox{d}q-\frac{\partial^2 \psi}{\partial x\partial y}\mbox{ d}x}{\mbox{d}y}$$

Substituting these back into our original second-order PDE (equation 10.8) yields

$$A\left(\frac{\mbox{d}p-\frac{\partial^2 \psi}{\partial x\partial y}\mbox{ d}y}{\mbox{d}x}\right) + B\frac{\partial^2 \psi}{\partial x \partial y}+C\left(\frac{\mbox{d}q-\frac{\partial^2 \psi}{\partial x\partial y}\mbox{ d}x}{\mbox{d}y}\right)
= f\left(x,y,\psi,\frac{\partial \psi}{\partial x}, \frac{\partial \psi}{\partial y}\right).$$

If we multiply both sides by $\mbox{d}x\mbox{d}y$, we can cancel the denominator and be left with

$$A\left(\mbox{d}p\mbox{d}y-\frac{\partial^2 \psi}{\partial x\partial y}(\mbox{d}y)^2\right)
+ B\frac{\partial^2 \psi}{\partial x \partial y}\mbox{d}x\mbox{d}y+C\left(\mbox{d}q\mbox{d}x-\frac{\partial^2 \psi}{\partial x\partial y}(\mbox{d}x)^2\right)
= f\left(x,y,\psi,\frac{\partial \psi}{\partial x}, \frac{\partial \psi}{\partial y}\right)\mbox{d}x\mbox{d}y,$$

which we can rearrange as
$$ A\mbox{d}p \mbox{d}y + C\mbox{d}q\mbox{d}x
- f()\mbox{d}x\mbox{d}y - \frac{\partial^2 \psi}{\partial x \partial y}\left(A(\mbox{d}y)^2-B\mbox{d}x\mbox{d}y + C(\mbox{d}x)^2\right)=0.$$

To extract the first-order ODEs for this system, we need to remove the remaining partial derivative $\frac{\partial^2 \psi}{\partial x \partial y}$, recognising that the characteristic lines will only exist when the coefficient in front of this partial derivative is zero, i.e., when
$$A(\mbox{d}y)^2-B\mbox{d}x\mbox{d}y + C(\mbox{d}x)^2= 0.$$

Dividing through by $\mbox{d}x$, we can express this coefficient as
$$A\left(\frac{\mbox{d}y}{\mbox{d}x}\right)^2-B\frac{\mbox{d}y}{\mbox{d}x} + C= 0,$$

which is a quadratic equation for the path of the characteristic lines, such that we can solve for the gradient by applying the quadratic formula,

$$\frac{\mbox{d}y}{\mbox{d}x} = \frac{B\pm\sqrt{B^2-4AC}}{2A}=\frac{B\pm\sqrt{\Delta}}{2A},$$

where $\Delta$ is the *discriminant*.

From here you can see how the *discriminant* relates to the existence of real solutions to the quadratic and hence the existence of characteristic paths.

*   If $\Delta > 0$ then there are two real solutions and hence two characteristic paths - a **hyperbolic** equation.
*   If $\Delta = 0$ there is only one real solution and hence a single characteristic path - **parabolic**.
*   If $\Delta < 0$ then the characteristic paths are complex and there is no path along which the problem can be reduced to an ODE - **elliptic**.

If there are no characteristic paths, then a disturbance at one point in the domain doesn't follow any discrete paths, but instead influences all points in the solution domain.

## 📂 10.4.4 Impact of PDE Classification on its Numerical Solution

---

In practice, most **parabolic** and **hyperbolic** equations will have one independent variable that behaves like time and others like space. In such cases, the solution will depend on the initial values. These equations involve a temporal evolution of the solution, e.g., $u(t,x,y,z)$. They are *initial value* or *Cauchy* problems.

Looking at the Heat Diffusion and Wave Equation, we see that if we know the value of $u(t=0,x,y,z)$ everywhere at some time, the right-hand side of the equation tells us how the flow will evolve.

An **elliptic** problem is more like a steady-state solution to a problem in space, e.g., $u(x,y,z)$, where you aim to solve for the internal points in the domain that will satisfy the governing PDE and the supplied boundary conditions. In general, it is not possible to simply integrate from the boundary the way you would do in an initial value problem, as changes in the later part of the domain will impact values in the earlier part due to the lack of any characteristic direction and the propagation of disturbances in all directions.

In most cases, the differences in the solution methods between **parabolic** and **hyperbolic** equations are subtle, and many systems may alternate between these two states. The difference between an **initial value** PDE and a **boundary value** PDE is far more significant.

### 🔹 Disturbance Propagation in Hyperbolic and Parabolic Equations

While ultimately the methods to solve these are similar, the existence of singular versus multiple characteristic lines does impact how disturbances propagate within the domain.

In a **parabolic** equation, like thermal or chemical diffusion, a change at one point in the domain has an infinite speed of propagation, meaning it is felt through all subsequent time steps and almost the entire domain instantly, though the effect decays rapidly with distance. They feature a smoothing property where sharp initial jumps or singularities disappear and become smooth curves as time progresses.

In a **hyperbolic** equation, like the Wave Equation, a disturbance only affects specific regions bounded by characteristic curves (like a light cone or sound barrier). Sharp jumps or shocks in the initial data do not smooth out;

Supersonic flows with shocks are a great example of this. Pressure fluctuations travel at the speed of sound and get trapped in a Mach cone. The edges of this cone are the characteristic curves which isolate the rest of the domain from disturbances in this region.
