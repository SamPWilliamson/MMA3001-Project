GitHub is a cloud-based platform where you can store, share, and work together with others to develop and maintain code. - You can use this to store word documents and things as well (just not huge data files - unless you are willing to pay).

Some key terms here to be aware of:

> ### 📖 **Repository**:
>where you store your source code files (.cpp, .py, .ipynb). This can be hosted in the cloud or via a company's private servers for security.

> ### 📖 **Branch**:
> If you think of your code growing like a tree, then a branch is a copy of the code that splits off from the 'Main' code. This will be a perfect copy of the code at the instant the branch was created, which can now be changed independently of the main branch.

To understand branching, consider the example in figure 1.2.1, below. Each circle reflects a change made to the code. If a user wants to add feature A, then rather than risk damaging the entire code and rendering in it useless until the modification are implemented and tested, they instead make a copy of the code, <font color='blue'>blue branch</font>, and create a separate version that can be developed. Once everyone involved is happy with the changes, this branch can *merged* back into the *main* branch of the code.

>### 📖 **Merge**
>removes a branch and replaces the parent branch with the one that is being removed.

From the example in Figure 1.2.1, we can see that at one point <font color='green'>Feature A-1 branch</font> is merged back into the <font color='blue'>Feature A branch</font>, before <font color='blue'>Feature A branch</font> is merged back into the <font color='gray'>Main branch</font>.

One of the key features of GitHub is that you can realtively easily see how files and code differs between branches, as well as who made the edits and when they were made.

> ### 📖 **Pull**:
> is when you call a version of the code from the *repository* to your local machine (or Colab runtime).

Let's say you want to work on the <font color='blue'>Feature A branch</font>, you would *pull* this branch to your machine or Colab notebook and replace the current version of the files on your machine. This won't replace all files, like a dumb copy, but instead tracks what is different between the files in each branche.

> ### 📖 **Push**
 is the opposite. This is when you send the current version you are working on back to the *repository*.

> ### 📖 **Commit**
>is when you let the repository know that you have made a substantial change.

A *commit* and *push* work as follows. We don't want to see every keystroke as a new code is added, one letter at a time. Instead, when you want to record a change you *commit* it to the branch, with a comment to explain the change, e.g. 'changed random number generater function'. Only *committed* changes can be *pushed* back to the repository. If you try to *push* a change and GitHub tells you that there is no change to make, then you have probably forgotten to *commit* your changes.

By working with GitHub, you will get a running record of all the changes that have been made to your project, when they were made, what changed, who changed it, and why (if people have added comments to their commit statements).

This is particularly useful for collaborative development, as all users can *pull* the latest changes made by others and ensure they are working on the same version, whether they are working remotely or across multiple computers at home and in the office. ### 🔹 Cloning a GitHub Repository

> ### 📖 Cloning (in GitHub)
is when you take a copy of the existing code and files in a repository, and can then edit it without any of your changes ever directly impacting the original repository.

You can clone a public repository directly into your Colab environment using `!git clone` (remember the '!' is just used on tell Colab to run this a linux shell command). This will create a local copy of the repository's contents.

```python
# Example: Clone a sample repository
!git clone https://github.com/googlecolab/colabtools.git

# List the contents of the cloned repository
!ls colabtools
```
> ### 🧩 Code Explanation
> `git clone https://*` tells git to make a fresh, independent copy of the code that is stored by the repository at the provided web address. This can be a github.com hosted site or a private GitHub server.
>
> By default, the code will be copied ononto to your machine in the local directory, contained in a folder with a title that corresponds to the *.git* file name.
>
> `ls colabtools` is the linux command to list all the files in the 'colab-samples' directory. Like `dir` if you are a DOS person.

We will look at an example of this in a full video a little later.

After cloning, you'll often want to change your current directory to work within the repository. You can do this using the `%cd` magic command.


> ### 🧩 Code Explanation
> `%` in front of a command in Colab makes that a **magic** command. Each magic commands is specific in purpose.
> `%cd` tells Colab to use the linux (same in DOS) change directory command to change the local directory. The % ensure that this change is not specific to this cell, but will persist and change the local directory in all subsequent cells.

### 🧪 Adding to a Repository

Files can be added to a repository in multiple ways.
* Through your web browser, you can navigate to the main GitHub page of your repository and click on the *Add file* button and either create or upload files.
* You can create a file in the Colab Notebook by navigating to the Files Menu, clicking on the three dots that appears when you mouse over a folder and selecting new file
* Or executing OS commands in colab like **touch**

> ### 📖 touch
> is a linux command that update the last time a file was accessed. The system just reaches out an touches the file without changing it. If the file doesn't exist then a `touch` directed at that filename will create it.

If you create a file through GitHub, you will be required to **commit** this change at the time of creation. When operating on a local machine, you need to manually tell GitHub to when to **commit** changes to the repository.

Let's take a look at an example.

Let create a basic Python code and add it to a new file. Let's us Gemini for a basic 'Hello world' python code.

> ### 🗣️ AI Prompts
> 'create a hello world python code'.

> ### 🧠 AI Output - Gemini (June 2026)
> code in following cell

````python
print("Hello, World!")
````

Ok, so this is some basic, we needed bother using AI for it, but what the hell. So how do we get this into a repository from a code cell? For the sake of this example, I don't want to look a getting a Jupyter notebook into a repository, but rather get working python code in there.

> ### 🧾 Instructions
> 1. Create new file called *hello.py*, using the `touch` command - run the cell below.



````python
!touch hello.py
````

This file should now appear in the *File Menu* within the project directory.

You can access the file from the *File Menu* by double clicking on it, and it should open in a window on the right side of your browser.

> 2. Paste the code `print("Hello, World!")` into the hello.py file
3. Run the code from the hello.py file as follows.

````python
# run hello python script
!python hello.py
````

> 4. Tell GitHub that we want to include this file.

````python
!git add hello.py
````

If you navigate back to the GitHub page you will see that nothing has changed. By `add` we told our local instance of the repository that this file exists, but we haven't told it that we want to **commit** this as an update to the repository.

> 5. **commit** the change using the `git commit` with `-m` flag to add a note that will be stored to explain this commit.

````python
!git commit -m "added the hello file"
````

> ### ⚠️ Point of note
> If you have followed the instructions above you should have got an error message `Author identity unknown ...`. Remember who we said that a Git Repository was create for collaboration? One reason for this is that it doesn't let people make anonymous changes to a code.

As a data repository it will keep a record of changes and who made them, so we need to tell it who we are. We can do this at any time and it will be stored in our .git files.

> 6. Use `git config` to tell the repository who you are, e.g.

````python
!git config --global user.email "callum.atkinson@monash.edu"
!git config --global user.name "Callum"
````

> 7. Try to commit again,

````python
!git commit -m "added the hello world file"
````

We can see a list of changes we have made by using the `git status` command. This is not a necessary step, but handy to be able to see what will be updated when we send the local version back to the cloud.

````python
!git status
````

> ### 📖 **git status**
> shows a list of changes that have been made to all watch files in the direct folder and subdirectories, since the last **commit**.

As we have committed the addition of the new file we have no new changes to report.

If you look at the online version of the repository, you will see that *no changes* have been made. To do this we need to **Push** the changes.

> 8. Push the current committed version to the GitHub servers using `git push`.

````python
!git push origin master
````

> ### 🧩 Code Explanation
> the `push` commands require us to specify the location we wish to push to, and the branch. We previously defined *origin* with the GitHub URL when we ran the `git remote add origin` commands, so we can just refer to it as *origin* now.

> ### 💬 **Note:** GitHub create a default **main** branch, but locally the default was **master**. This is legacy issue as in early versions the deafult main branch was master. If you go to GitHub you will now see that your project has both a main and a master branch.

If you want to, you can clean the repository up a little by renaming the *master* branch to main* using the `git branch` command. We first remain our current branch, then **push** it as *main*.


````python
# Rename present master branch to main
!git branch -m main

# push as main
!git push origin main
````
