## 📂 1.4.3 Safe Use of AI and Protecting Your Data

Remember that *'nothing comes from nothing!'* and *`there's no such thing as a free lunch'*. When using AI tools, it is important to be aware of the risks - who owns the inputs, and what rights you have over the outputs. Many AI tools are learning from your inputs and responses. Learning from you and potentially learning about you.

The use of these tools involves multiple data risks, which can impose severe threats to your privacy, security, and data integrity. These can consist of

* **Data Privacy and Data Leaks**: Sensitive data (personal or corporate secrets) entered into public generative AI tools can be retained, used for retraining, and have the potential to be leaked in future outputs (even if unintentionally).
* **Model Inversion and Inference**: Adversaries can use the outputs of a model to reverse-engineer or infer the private data used for training.
* **Data Poisoning and Manipulation**: Attackers can corrupt training datasets, causing models to make incorrect decisions, misclassify data, or create security backdoors.

The latter is more about being careful with AI responses as these have the potential to be manipulated, similar to trusting online reviews that come from paid influencers or negative reviews from competitors. The following video will help explain some of these risks.

### 🔹 Using Publicly Avaliable AI Tools

If you are using a public AI tool, it is important that you check the user agreements you actively or implicity agree to when using that tool. Pay particular attention to:
* Whether you or the AI platform has ownership rights over the generated outputs.
* What can the AI platform do with the information you give it?
* What data security and identity protection is provided by the AI platform?

> ### ⚠️ Point of note
> Always consider who owns what you upload and generate and what information is the AI platform storing about you.
