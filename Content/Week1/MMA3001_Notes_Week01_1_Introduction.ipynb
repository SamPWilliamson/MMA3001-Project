>
>The following Except from Google will help explain the process:
>
>Booting into "high memory" in the 90s was a critical, often frustrating, ritual for PC gamers trying to run demanding MS-DOS titles. Because DOS was designed to operate within a strict 640KB limit (conventional memory), gamers had to use advanced configuration techniques to move drivers, TSRs (Terminate and Stay Resident programs), and mouse/CD-ROM drivers into the upper memory area (640KB–1024KB) to free up space.
>Here is how "high memory" configuration was managed in the 90s:
>1. The Core Files: CONFIG.SYS and AUTOEXEC.BAT
Gamers spent hours editing two main text files to manage memory.
CONFIG.SYS: Loaded device drivers. This is where HIMEM.SYS (Extended Memory Manager) and EMM386.EXE (Expanded Memory Manager) were loaded to enable upper memory access.
AUTOEXEC.BAT: Executed upon startup to load drivers (like mice, CD-ROMs) and set environment variables.
2. Techniques to "Go High"
To free up conventional memory, drivers and TSRs were loaded "high" using specific commands:
DEVICEHIGH=... (in CONFIG.SYS): Loads drivers into Upper Memory Blocks (UMBs).
LH or LOADHIGH (in AUTOEXEC.BAT): Moves executable programs into upper memory.
DOS=HIGH,UMB: A crucial line in CONFIG.SYS that allowed DOS to load parts of itself into the High Memory Area (HMA) and managed UMBs.
3. Memory Managers and Automation
MemMaker (DOS 6.22): A built-in utility introduced in DOS 6.22 that automated this process. It would analyze the system, try different loading configurations, and automatically edit CONFIG.SYS/AUTOEXEC.BAT for optimal performance.
QEMM (Quarterdeck Expanded Memory Manager): A popular third-party tool often superior to MemMaker for freeing up maximum conventional memory.
4. Common Hurdles and "Memory Hell"
The 640KB Limit: Many games required 580KB–610KB of free conventional RAM to launch, causing "Not enough memory" errors.
TSR Conflicts: Loading a driver high could sometimes hang the system upon booting, requiring a reboot and safe mode (pressing F8).
Sound Card Drivers: Sound Blaster drivers were infamous for taking up precious conventional memory if not loaded correctly.
5. Alternative Boot Solutions
Boot Disks: If a game was too memory-intensive, users would create a "custom boot floppy" designed specifically for that game, bypassing normal startup routines.
Windows 95/98 Reboot in DOS: Later in the 90s, gamers used "Restart in MS-DOS Mode," which allowed Windows to unload itself, providing a clean DOS environment.
This manual optimization was essential until Windows 95/98 became prevalent and changed the memory management paradigm entirely.

### 🔹 Training and the Role of Engineers in Machine Learning



The *training* that is performed in ML is very different from programming. In programming, we have to consider the user input and come up with a algorithm, as series of steps the computer must use to arrive at an output. In ML, we provide a model with a number of internal parameters, and use known data to enable the model to adjust its internals until is best matches the known outputs. The process is generally something like this:

> ### 🧾 Training Procedure
>1. Gather a training dataset consisting of a collection of inputs to the model and the outputs we expect from the model for those inputs.
2. Select the type of model we want to train.
3. Train the model by presenting the training inputs and adjusting the model’s parameters when it gets the outputs wrong.
4. Repeat step 3 until we are satisfied with the model’s performance.
5. Use the now-trained model to produce outputs

>from Kneusel, R. T. (2023)

### 🔹 Binary Transistors States and Storage

Firstly, it is important to note that all computers operate on *Machine Code*, i.e. as set of instructions that a computer's central processing unit (CPU) can understand and execute directly. This is represented by a series of 0, 1, or binary digits. This corresponds to the on / off state of the billions of transistors that sit on a CPU.

Each binary unit or binary digit (0 or 1) is called a bit (for short), which is the smallest unit of measurement used to quantify computer data. It contains a single binary value of 0 or 1. We can use a single bit to store the number 0 or 1 (pretty limited for a number) or more often use it to store a BOOLEAN value (True == 1, False == 0).

To understand how computers handle larger numbers we need to consider how to present numbers from our base 10 decimal system in base 2 binary.

> ### 🧪 Example
>
> Consider the number 123 in decimal, which can be denoted $(123)_{10}$, the subscript indicating that the number is in base-10.
>
>In base-10, each integer is raised to the power of ten such that we can expand this as
>
>$1(10^2) + 2(10^1) + 3(10^0) = (123)_{10}$
>
> To convert to base 2, we need to consider what this would look like if it were represented as a sum of powers of two, or $\sum^N_{n=0} a_n(2^{n})$, where each value of $a$ can be 0 or 1.
>
> To do this we divide by lots of 2 until we get to the final digit. Remember, the digits in base 10 are remainders of 10, as these binary digits are remainders of 2.
> \begin{align}
123 / 2 = 61 &\mbox{ remainder } \bf{1}& (\mbox{last digit})\\
61 /2 = 30 &\mbox{ remainder } \bf{1}&\\
30/2 = 15 &\mbox{ remainder } \bf{0}&\\
15/2 = 7 &\mbox{ remainder } \bf{1}&\\
7/2 = 3 &\mbox{ remainder } \bf{1}&\\
3/2 = 1 &\mbox{ remainder } \bf{1}&\\
1/2 = 0 &\mbox{ remainder } \bf{1}& (\mbox{first digit})
\end{align}
> Hence $(123)_{10} = (1111011)_2$
>
> As an 8-bit integer this would be represented as 01111011.

Based on the example above, we can see that storing the number $123$ requires at least 7 bits of data. The smallest varible we usually define for an integer (i.e. non-decimal number) is an 8-bit integer (**ushort8** in C/C++ or **uint8** in Python) which coresponds to 1 *Byte*. The 'u' stands for *unsigned*, all values are positive.

Digital data is measured in units of increasing magnitude: bits, bytes, kilobytes (KB), megabytes (MB), and gigabytes (GB).

*So a megabtye is $10^6$ bytes?* - **It depends who you ask!**

This is still a contentious area in computing. Have you ever bought a 16 GB USB Drive, connected it to your computer and found out is only 14.90 GB?

> WHAT?

The reason for this is that file systems are binary (base-2), so once values started getting over $2^{10} = 1024$ Bytes, people wanted to express is like we would for base-10, i.e. in kilo, mega, giga units. In binary, the closest power of 2 to 1000 is 1024, so people started refering to 1024 Bytes as 1 kilobyte (KB), and 1024 kilobytes as 1 megabyte (MB). This lead to confusion as we humans typically use the decimal system were kilo is 1000 units. So a hard disk manufacture might sell you a solid state drive (SSD) packaged as 500 GB, which Windows might tell you was 465 GB.

> ### 💬 Memory Confusion

### 🧪  Interpreted Language Example in Python

An *interpreted language* does away with the explicit compilation step, which can be a real pain as your code grows to span multiples files and libraries.

The Python executable, which you either call from the commandline or run through Integrated Development Environment (IDE) software like [Anaconda's Spycommand-line or run through Integrated Development Environment (IDE) software like [Anaconda's Spyder](https://www.anaconda.com/docs/getting-started/guides/ides/spyder), [Visual Studio Code](https://code.visualstudio.com/) or in this case through the Google Colab environment, acts as an interpreter. It moves through your code line-by-line and executes each command as it comes to it.

**But isn't this what at compiler does?**

A compiler is different in that it goes through the entire code before execution (not just line-by-line), working out what the program is trying to do and optimisers the order of things like memory handling, to try and get better performance. This *compile* step takes time, but enables the final code to better adapt to the local computer architecture and hence get better performance.

Traditionally, *interpreted* code are much slower, as they need to be more conservative as they don't know that the next line of code will ask it to do.

e.g. in Python we can set a variable without telling it what we are going to use it for.
```python
a = 12
a = 2*a + 1.4
```
In this case, we are assigning the variable *a* an integer value, then in the next line adding a floating point (decimal). We might not care if a number is 2 or 2.0, but a computer does at a hardware level. These are two separate data type (an *integer* and a *float*). Not knowing what comes next, an interpreted language needs to handle different cases and make changes to the data at the machine-level depending on the value of 'a'. Interpreted language tend to use more memory and make more memory operation as they convert data between types as needed.

>### 🧾 Instructions
> Run the following cell of Python code and note the output.

````python
import sys

a = (2)
print(type(a))
print(f"Size of integer: {sys.getsizeof(a)} bytes")

a = 2*a + 1.4
print(type(a))
print(f"Size of integer: {sys.getsizeof(a)} bytes")
````

> ### 🧩 Code Explanation
>* Line 3: sets a to the value 2.
>* Line 4: prints the data type of the varible a to the output stream.
>* Line 5: prints the data size of variable a.
>
> Line 7: adds a float (decimal value) to a
> Line 8,9 print the type and size of a after the operation on line 7.


If your have been paying attension (I am sure you all have), you are probably asking 'Why does Python need 28 bytes of (i.e. 28x8 = 224-bits) for a single integer. Well this is another downside to interpreted languages. The flexibility it offers for our coding come with additional memory overhead. In an a compiled code like C++, we would have to explicity define the type of each variable before we used it.

Using inbuilt commands and libraries in Python (which are precompiled) can reduce this gap in performance, as can just-in-time compilation used in languages like Julia.
