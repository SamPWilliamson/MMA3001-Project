## 📂 1.5.1 Differences Between Testing, Debuggind and Validation

---

Testing, Debugging and Validation are all key and complementary processes in the development of engineering and software tools.

> ### 📖 Definition
> **Testing** is the process of checking that your code (and subroutines) works as intended and identifying potential problems.
> **Debugging** is the process of finding the root of the problem within the logic or operation of the code and correcting it.
> **Validation** is confirming that what you have developed meets the original requirements and provides the solution that motivated the codes development. This includes evaluating the performance under realistic conditions.

Development of a code should go through all of these phases, at varying levels. Each function that is developed needs to be *tested*. Failure of a test will require *debugging*, with a final *validation* needed once all the components have been assembled.

## 📂 1.5.3 How Can I Efficiently Test My Code?

---

At this point, you are probably looking at the additional **nine** functions we generated to test that one little code and thinking, even if I use AI to generate these, this is going to fill my code with distracting tests functions.

As with most things, your are not the first person to encounter this and someone has already come up with a better way to do this.

**Pytest** (https://docs.pytest.org/en/stable/) is a Python library that has been developed to make it easier to write, run and collate the results of small to large scale complex functional testing for applications and libraries. This allows you to run a single command and have *pytest* automatically search your project directory for all 'test\_\*.py' or '\*\_test.py' files, and automatically run and report on the outcome of all 'test\_\*()' functions. This allows you to place all test related functions in a seperate directory, removed from the basic source code, and be able to re-run these to check that any code modification have not introduced unexpected behaviours.

> ### ⚠️ Point of note
> You have to create separate test functions with `test` in the filename to get the benefits of using this tool.

In addition to ensuring all test functions are prefaced with 'test_' in their name, the *assert* command is used.

> ### 📖 Definition
> **assert**: in Python is a debugging/testing command used to test if a condition is true. Unlike a normal 'if' statement, this automatically raises an AssertionError exception

### 🧪 Example - Pytest Positive Test Automation

Pytest will automatically look for all **AssertionError** outcomes and provide your with a report.

A pytest compatible code, looks like this


````python
%%writefile test_sample.py
"""sample pytest code test_sample.py"""

def func(x):
    return x + 1

def test_answer1():
    assert func(1) == 3

def test_answer2():
    assert func(3.1) == 4.1

````

> ### 🧩 Code Explanation
> `%%writefile test_sample.py` is one of the Colab *magic commands* which tells Colab to write all the code in this cell into the specified filename *test_sample.py* when the cell is executed.
>
> In this case I have include the function we wish to test `def func(x)` in this file, but in practice you would have the function in a separate file to separate your working code form the *test\_\*.py* files.
>
> `assert` replaces the more cumbersome series of *'try/except'* and print statements. A single  *'assert'* returns the *True/False* condition and outputs messages.
>
> `assert age >= 18, "User must be at least 18 years old."` will add a message to the test so help us keep track of the condition we are testing.

The **beauty** of this is that **Pytest** will handle the test reporting for us.

This done by running the `pytest` on the linux backend.

> ### ⚠️ Point of note
> Pytest searches and run test in all *'test_*' or '*_test' files, so no need to specifically point it to our *.py* file. Infact is can run all tests in the project directory with this one command.

````python
# run pytest
!pytest
````

Pytest produces a relatively well formated report that list the results of the tests that were run in each file, grouped in the passes and fails. The [100\%] next to the filename indicates the percentage of tests in that file that were **completed** (pass and fail).

Output messages are only displayed for the failed tests. In this case, **test_answer1**.

> ### ⚠️ Point of note
 Pytest will stop at the first unsuccessful `assert` in a function. If you want to ensure Pytest runs all tests, you need to create a separate 'test_' **function** for each test. They CAN be in the same *.py* file.

## 📂 1.5.4 Asserts and Error Handling for Floating Point Numbers

---

When dealing with floating point numbers, truncations and rounding errors can lead to unexpected inequality. This should make more sense when we discuss numerical precision in Week 2, but for now just note that a computer looks at 6.000001 and 6.000000 and does not see these as equal.



### 🧪 Example - Pytest with Floating Point Values

Consider the following example where we have a function that simply adds 0.2+0.2+0.2 and then checks if this is equal to 0.6. Easy right?

> ### 🧾 Instructions
> 1. run the first cell below to generate the test_sample.py file with the included function and two test cases.
> 2. Run `!pytest` to run the test functions and report on their success/failure.

````python
%%writefile test_sample.py
"""sample pytest code test_sample.py"""

import pytest # we need to load the pytest library to use pytest.approx()

# the function we wish to test
def func():
    return 0.2 + 0.2 + 0.2

# a simple assert test - output should = 0.6
def test_sum():
    assert func() == 0.6

# a test to check the value is close enough
def test_float():
    assert func() == pytest.approx(0.6)
````

````python
!pytest
````

The first test here has **FAILED** as the sum of 0.2 + 0.2 + 0.2 was not exactly equal to 0.6, but rather $0.6 + 10^{-16}$. For all intensive purposes, they are the same, but not exactly, due to roundoff error in floats (*reminder we will explain why this happens in Week 2.*)

We can avoid these floating-point roundoffs throwing errors by using the **pytest.approx()** function when comparing float values. This second test_float() did not throw an error during testing.

> ### 🧩 Code Explanation
> `pytest.approx()` tests number with a default relative tolerance of $1\times 10^{-6}$. Equivalent to single precision accuracy. You can manually set a different tolerance as follows
>
> `assert value == pytest.approx(expected, rel=1e-12)`
>
> will set a relative tolerance of $1\times 10^{-12}$

### 🧪 Example - Negative Test Assertion in Pytest

*Pytest* can also be used with **negative** tests, using `pytest.raises(Exception)`. For instance consider a function that divides one value by the other.

````python
%%writefile test_sample2.py
"""sample pytest code test_sample2.py"""

import pytest

# the function we wish to test
def func(a,b):
    return a/b

# a simple positive assert test - we should get 10/2=5
def test_integer():
    assert func(10,2) == 5

# a positive assert test for a float
def test_float():
    assert func(10.0,2.0) == pytest.approx(5)

# a negative test for division by zero
def test_zero():
    func(10,0)

# a negative test for division by zero
def test_zero2():
  with pytest.raises(ZeroDivisionError) as excinfo:
    func(10,0)

  assert "division by zero" in str(excinfo.value)
````

````python
!pytest
````

In this case we have two positive tests that pass, and two negative tests. The first of these fails as it raises and an error, the second negative test passes, as we are expecting it to raise a ZeroDivisionError. If it didn't we know something has gone wrong and it isn't handling division by zero as we expect.

## 📂 1.5.5 What are Software Bugs?

---

Bugs are an unavoidable part of all software and code. Most of the updates you get for your phone, computer, etc. will focus on fixing bugs or security vulnerabities.

> ### 📖 Definition
>**Bug** is a software error that causes unintended results, erroneous calculations, or unstable behaviour that might cause a code to crash.
>
>**Security Vulnerability** is a specific type of bug that can be exploited by malicious actors to gain unintended access to a system, data, or resources that compromise a system.

We won't be focusing on *vulnerabilites*, but in this unit, we will be encountering situations where numerical methods become unstable, and no you will definitely introduce a bug into all but the most basic codes.

Bugs can take many forms. The common ones that you will encounter are:
> ### 📖 Definition
>* **Functional Bugs**: Features that do not work as intended, e.g. outputs routines that don't overwrite existing files when you expect them to.
> * **Logical Bugs:** Errors in logic or calculations that lead to incorrect results. For example, returning the minimum instead of the maximum value.
>* **Syntax Bugs:** Errors in the code, such as missing semicolons, improper indentation, or typos. Runtime compiler should throw errors for these.

The above Bugs are common, but also fairly easy to identify and fix.

The next level of bugs you will encounter are often harder to recognise. They consist of
> ### 📖 Definition
>* **Performance Bugs**: Problems that cause a code to fail or become inefficient for larger problem sizes. E.g. works great on small test cases by fails on production size problems.
> * **Boundary/Edge Case Bugs**: Errors that occur when input values are at the extreme limits of the allowed range. This could be caused by negative values that aren't handled by the code, or overflow of variables due to larger values.

> ### 💬 Some Bugs Appear with Changes In Data Size
> I ran into a performance bug during my PhD. I developed all my code for 1000 x 1000 pixel images, then went to use a camera with 4000 x 2400 pixel and could no longer fit some values in memory.

So of these error will be obvious, but other will be subtle, with boundary cases, leading to logic errors that may cause the program to crash and alert you to the problem, or others that might influence the calculation but still allow the code to operate in what appears to be the *normal* fashion. While the latter is generally harder to detect, in both cases it can be difficult to find the source of the error.

> ### 💬 Some Bugs are Hardware Specific.
Some functions and background memory handling may change behaviour depending on your machine. I have had bugs in codes that had no impact on the results when run on a MAC, but caused Segmentation Faults on a Super Computer - that was a REAL PAIN to debug.

If your code returns an error during runtime, i.e. not a syntax error, there is a good chance it is a **Segmentation Fault**.

> ### 📖 Definition
 **Segmentation Fault** - often abbreviated to *Segfault*, is an error that occurs when a program attempts to access a memory location that it is not permitted to access, or attempts to access an allowed memory location in a way that is not permitted. The best example of this is trying access an index that is out of range, e.g. 'x = y[1000]', where 'y' index ranges from 0 to 999.

### 🧪 pdb Debugging a Python Code

Consider the following code, which is intended to return all values from the beginning of the list to a specified end value.

````python
%%writefile buggy_indexing.py

def get_sublist_up_to_index(data_list, end_index):
    """Return a sublist from the start of `data_list` up to `end_index`.

    This function returns all elements from index 0 up to,
    `end_index`. It validates that `end_index` falls within the valid index
    range for `data_list`.

    Args:
        data_list (list): The list from which to extract a sublist.
        end_index (int): The ending index for the sublist.
            Must satisfy ``0 <= end_index < len(data_list)``.

    Returns:
        list: A sublist containing elements ``data_list[0:end_index]``.

    Raises:
        IndexError: If `end_index` is outside the bounds of `data_list`.
    """
    # this an input validation text to end_index is within bounds of data_list
    if not (0 <= end_index < len(data_list)):
        raise IndexError("end_index is out of bounds for the list.")

    return data_list[0:end_index]
````

> ### 🧩 Code Explanation
> We have included
>
> `raise IndexError("end_index is out of bounds for the list.")`
>
> to give use control of the behaviour of the code when it encounters a value that is outside of range. Rather than simply throwing a default error, Python will instead use the custom error message we have provided.

I have used the Gemini 2.5 Flash AI (June 2026) to generate a series of **positive test** functions that will check that the expected values are returned, along with **negative test** function that will check we get the expect out of range index errors.

> ### 🧾 Instructions

> 2. Run the cell below to generate the associated test_\*.py file.

````python
%%writefile test_buggy_indexing.py
"""
Pytest Compatible Test Functions for buggy_indexing.py
"""
import pytest

# import the functions form buggy_indexing.py as a module
import buggy_indexing

def test_get_sublist_correct_case():
    my_list = [10, 20, 30, 40, 50]
    # Intended behavior: get elements up to index 2 (10, 20, 30)
    assert buggy_indexing.get_sublist_up_to_index(my_list, 2) == [10, 20, 30]

def test_get_sublist_another_correct_case():
    my_list = ['a', 'b', 'c']
    # Intended behavior: get elements up to index 0 ('a')
    assert buggy_indexing.get_sublist_up_to_index(my_list, 0) == ['a']

def test_get_sublist_full_list():
    my_list = [1, 2, 3]
    # Intended behavior: get elements up to index 2 (1, 2, 3)
    assert buggy_indexing.get_sublist_up_to_index(my_list, len(my_list) - 1) == [1, 2, 3]

def test_get_sublist_index_error_handling():
    my_list = [1, 2, 3]
    with pytest.raises(IndexError):
        buggy_indexing.get_sublist_up_to_index(my_list, 3)

def test_get_sublist_index_error_handling2():
    my_list = [1, 2, 3]
    with pytest.raises(IndexError):
        buggy_indexing.get_sublist_up_to_index(my_list, -1)
````

### 🔹  Explanation of the Indexing Logic Bug

If we ask Gemini AI provide to explain the bug we get the following explaination for the present bug, which is consistent with what we found.

> ### 🧠 AI Output
> **The Buggy Function: `get_sublist_up_to_index`**
>
>*   **Intention:** This function is designed to return a sublist containing elements from the start of `data_list` up to, and *including*, the element at `end_index`.
>*   **The Logic Bug:** The line `return data_list[0:end_index]` is the source of the problem. In Python, list slicing `data_list[start:end]` is *exclusive* of the `end` index. This means `data_list[0:end_index]` will return elements from index `0` up to `end_index - 1`, thereby *excluding* the element at `end_index`. This is a classic off-by-one error in indexing logic.
>*   **Impact:** The function won't raise an `IndexError` as long as `end_index` itself is within the valid range for slicing (i.e., `0 <= end_index <= len(data_list)`), but the *result* will be incorrect—it will always be one element shorter than intended.
